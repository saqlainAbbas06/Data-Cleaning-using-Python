**CLEAN THE DATA**

**Task 06: Remove duplicate StudentIDs**

```python
clean_students = students_df.drop_duplicates(subset='StudentID')
clean_students
```

```python
clean_students.duplicated().sum()
```

**Task 07: Standardize Gender**

```python
clean_students['Gender'].astype(str).str.strip().str.title()
clean_students['Gender'].replace({'M': 'Male', 'F': 'Female'}, inplace=True)
```

```python
clean_students
```

**Task 08: Standardize Department**

```python
clean_students["Department"]=clean_students['Department'].str.upper()
```

```python
clean_students
```

**Task 09: Handle Missing Age**

```python
clean_students['Age'] = clean_students['Age'].fillna(clean_students['Age'].median())
```

```python
clean_students
```

**Task 10: Clean missing quiz marks**

```python
clean_results = results_df.copy()
```

```python
clean_results['Quiz1'].fillna(clean_results['Quiz1'].mean(), inplace=True)
clean_results['Quiz2'].fillna(clean_results['Quiz2'].mean(), inplace=True)
```

```python
clean_results
```

**Task 11: Correct invalid attendance**

```python
clean_attendance = attendance_df.copy()
```

```python
# Decision: If ClassesAttended > ClassesHeld, cap Attended at the maximum possible ClassesHeld.
# This assumes perfect attendance with a data-entry typo rather than wiping out the row.
invalid_attendance = clean_attendance['ClassesAttended'] > clean_attendance['ClassesHeld']
clean_attendance.loc[invalid_attendance, 'ClassesAttended'] = clean_attendance.loc[invalid_attendance, 'ClassesHeld']
```

```python
clean_attendance
```

**CREATE USEFUL FEATURE**

**Task 12: Attendance percentage**

```python
clean_attendance['AttendancePercentage'] = (clean_attendance['ClassesAttended'] / clean_attendance['ClassesHeld']) * 100

clean_attendance['AttendancePercentage'] = clean_attendance['AttendancePercentage'].round(2)

print(clean_attendance.head())
```

**Task 13: Assessment totals**

```python
clean_results
```

```python
clean_results['QuizTotal'] = clean_results['Quiz1'] + clean_results['Quiz2']
clean_results['TotalMarks'] = clean_results['QuizTotal'] + clean_results['Assignment'] + clean_results['Midterm'] + clean_results['Final']
clean_results['Percentage'] = (clean_results['TotalMarks'] / 100) * 100
clean_results.head()
```

**Task 14 - Performance category**

```python
def categorize_performance(percentage):
    if percentage >= 85:
        return 'Excellent'
    elif 70 <= percentage < 85:
        return 'Good'
    elif 60 <= percentage < 70:
        return 'Average'
    else:
        return 'Needs Improvement'

clean_results['Performance'] = clean_results['Percentage'].apply(categorize_performance)
clean_results.head()
```

**Task 15 - Export your cleaned files**

```python
clean_students.to_csv('students_clean.csv', index=False)
clean_attendance.to_csv('attendance_clean.csv', index=False)
clean_results.to_csv('results_clean.csv', index=False)

print("Cleaned datasets exported successfully!")
```

**Independent Challenge**

```python
def get_data_quality_metrics(df, df_name):
    rows = df.shape[0]
    columns = df.shape[1]
    missing_values = df.isnull().sum().sum()
    duplicate_ids = df.duplicated(subset=['StudentID']).sum() if 'StudentID' in df.columns else 0
    return {
        'Dataset': df_name,
        'Rows': rows,
        'Columns': columns,
        'MissingValues': missing_values,
        'DuplicateIDs': duplicate_ids
    }
# Get metrics for each cleaned dataset
students_quality = get_data_quality_metrics(clean_students, 'clean_students')
attendance_quality = get_data_quality_metrics(clean_attendance, 'clean_attendance')
results_quality = get_data_quality_metrics(clean_results, 'clean_results')
# Create a DataFrame from the metrics
data_quality_report = pd.DataFrame([
    students_quality,
    attendance_quality,
    results_quality
])
data_quality_report.to_csv('data_quality_report.csv', index=False)
print("Data quality report generated and exported as 'data_quality_report.csv':")
print(data_quality_report)
```
